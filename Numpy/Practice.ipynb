```python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])

a + b       # [11, 22, 33]
a - b       # [-9, -18, -27]
a * b       # [10, 40, 90]
b / a       # [10., 10., 10.]
a ** 2      # [1, 4, 9]
```
